This is the script I used to import my data using the MET API. I am including it here as plain text so that the code won't execute when I run and publish the notebook. 

```python
import requests
import json

url = "https://collectionapi.metmuseum.org/public/collection/v1/search"
params = {'q': 'woman'}
r = requests.get(url, params=params)
parsed = r.json()
ids = parsed['objectIDs']

objects = []
for i in ids[:500]:
    url = f"https://collectionapi.metmuseum.org/public/collection/v1/objects/{i}"
    r = requests.get(url)
    parsed = r.json()
    objects.append(parsed)

with open("woman_500.json", "w") as f: 
    json.dump(objects, f)
```

## filtering data
You can filter data through conditional statements using equivalence operator `==`

````python
# prints out a series that fulfills the condition

df['title'] == 'Woman'
````

````python
# saves the series to a variable

woman = df['title'] == 'Woman'
````

````python
# you have a series

type(woman)
````

````python
# and with df[] you have a dataframe

type(df[woman])
````

````python
df[woman].info()
````

```python
sentence = "Hello my name is Filipa and I teach python at Pratt Institute, School of Information."
```

```python
type(sentence)
```

```python
# we can tokenize our string

tokens = nltk.word_tokenize(sentence)
```

```python
# tokens are the basic building block for all text analysis tasks

tokens
```

```python
# NLTK comes with a part of speech tagger 

tagged = nltk.pos_tag(tokens)
```

```python
tagged
```

## checking out the `inaugural` corpus

```python
from nltk.corpus import inaugural
```

```python
inaugural.fileids()
```

```python
trump_2 = inaugural.words('2025-Trump.txt')
```

```python
# this speech isn't very long

len(trump_2)
```

```python
for i in inaugural.words('2025-Trump.txt')[:100]:
    print(i)
```

```python
# turn the corpus into an nltk.text.Text object

t_2 = nltk.Text(trump_2)
```

```python
# use dir() to see all the functions associated with this new object

dir(t_2)
```

```python
t_2.similar('nation')
```

```python
t_2.dispersion_plot(['nation'])
```

## checking out the `gutenberg` corpus

```python
from nltk.corpus import gutenberg
austen = gutenberg.words('austen-emma.txt')
```

```python
len(austen)
```

```python
a_text = nltk.Text(austen)
```

```python
len(a_text)
```

```python
a_text.similar('old')
```

```python
a_text.similar('young')
```

```python
a_text.similar('forest')
```

```python
a_text.dispersion_plot(['woman', 'man', 'happy', 'sad'])
```

```python
a_text.concordance('woman')
```

```python
# now, we can import nltk

import nltk
```

```python
sentence = "Hello my name is Filipa and I teach python at Pratt Institute, School of Information."
```

```python
type(sentence)
```

```python
# we can tokenize our string

tokens = nltk.word_tokenize(sentence)
```

```python
# tokens are the basic building block for all text analysis tasks

tokens
```

```python
# NLTK comes with a part of speech tagger 

tagged = nltk.pos_tag(tokens)
```

```python
tagged
```

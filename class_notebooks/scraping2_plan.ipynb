## looping through elements to extract data

```python
### looping through elements to pull out information about each one
#   - create a variable to hold the item you're looping through
#   - use find_all() to create the variable

articles = soup.find_all('div', class_ = 'result-card__info-wrapper') 

for i in articles[:3]: 
    print(i)
```

```python
# pull out: 
# - title
# - date 

## -> will need to layer the elements to get the date.

for i in articles[:3]: 
    print(i.a.text)
    print(i.p.span.text)
```

```python
### GROUP WORK!

# pull out:
# - author

for i in articles[:3]: 
    print(i.a.text)
    print(i.p.span.text)
    print(i.span.text)
```

## saving our data to lists

To review, here is the process for extracting elements from a webpage, now with the added step of saving to a CSV file:

multiple steps:
1. choose element that contains all data you want: titles, date, tags,...
2. note the html structure of data: which elements are contained within which elements?
3. write code that picks out the information we want from the data
4. process information into lists
5. save to CSV

```python
titles = []
dates = []
authors = []

for i in articles: 
    titles.append(i.a.text)
    if i.p.span is not None:
        date = i.p.span.text
    else:
        date = 'Nd'
    dates.append(date)
    authors.append(i.span.text)
```

```python
dates
```

## finally, saving to CSV

```python
import pandas as pd

df = pd.DataFrame({
    'date': dates,
    'title': titles,
    'author': authors
})
```

```python
df
```

```python
df.to_csv('heritage.csv')
```

```python
# first, create a soup of the next page

r = requests.get('https://www.heritage.org/big-tech?taxonomy_term_tid=143&page=1')
source = r.content
soup = BeautifulSoup(source, 'html.parser')
```

```python
# then, loop through the articles and save the info we want to lists

# be careful to change the list name here, you don't want to overwrite articles
articles1 = soup.find_all('div', class_ = 'result-card__info-wrapper') 

for i in articles1: 
    titles.append(i.a.text)
    if i.p.span is not None:
        date = i.p.span.text
    else:
        date = 'Nd'
    dates.append(date)
    authors.append(i.span.text)
```

```python
# re-create the dataframe

df = pd.DataFrame({
    'date': dates,
    'title': titles,
    'author': authors
})
```

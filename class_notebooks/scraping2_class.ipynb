```python
# loop through the articles and pull out title, author, date

# 1. create a list of articles containing all the data, i.e. articles = soup.find_all()
# 2. write a loop that goes through the articles variable, i.e. for i in articles: 
# 3. print out each element we want, i.e. print(i.a.text)

articles = soup.find_all('div', class_ = 'result-card__info-wrapper')

titles = []
authors = []
dates = []

for i in articles:
    titles.append(i.a.text) # getting title
    if i.find('p', class_ = 'result-card__date') is not None: # getting date
        if i.find('p', class_ = 'result-card__date').span is not None:
            date = i.find('p', class_ = 'result-card__date').span.text
        else:
            date = 'Nd'
    else:
        date = "Nd"
    dates.append(date)
    if i.find('span', class_ = 'result-card__link') is not None: # getting author
        author = i.find('span', class_ = 'result-card__link').text
    elif i.find('a', class_ = 'result-card__link') is not None:
        author = i.find('a', class_ = 'result-card__link').span.text
    else:
        author = 'No author'
    authors.append(author)
```

```python
dates
```

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    'title': titles,
    'author': authors,
    'date': dates
})
```

```python
df.to_csv('heritage.csv')
```
